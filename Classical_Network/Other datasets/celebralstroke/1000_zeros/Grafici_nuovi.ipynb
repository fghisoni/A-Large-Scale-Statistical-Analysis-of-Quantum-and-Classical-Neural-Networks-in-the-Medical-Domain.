```python
from matplotlib import pyplot as plt
import numpy as np
import ast
```

# Classical Neural Networks

```python
node_list = []
layer_list = []

with open('all_configs.txt', 'r') as f:
    for line in f.readlines():
        int_list = ast.literal_eval(line)
        layer_list.append(len(int_list))
        
        node_string = ''
        for i in int_list:
            node_string += str(i)
        node_list.append(node_string)

# layer_list[:10], node_list[:10]
```

```python
cnn_test_accs_layers_mean = []
cnn_test_accs_layers_std = []

for idx, node_string in enumerate(node_list):
    int_list = []
    for seed in range(1,11):
        fname = f'Results/Classical_network_0.7trainsize_testaccs_{seed}seed_{layer_list[idx]}layers_{node_string}nodes_Adam.txt'
        all_res_seed = []
        with open(fname,'r') as f:
            for line in f.readlines():
                all_res_seed.append(float(line))
            int_list.append(max(all_res_seed))
            # int_list.append(all_res_seed[-1])
    
    cnn_test_accs_layers_mean.append(np.mean(np.array(int_list)))
    cnn_test_accs_layers_std.append(np.std(np.array(int_list)))

index_max = cnn_test_accs_layers_mean.index(max(cnn_test_accs_layers_mean))
```

```python
dropout_test_accs_layers_mean = []
dropout_test_accs_layers_std = []

for idx, node_string in enumerate(node_list):
    int_list = []
    for seed in range(1,11):
        fname = f'Res_dropout/Classical_network_0.7trainsize_testaccs_{seed}seed_{layer_list[idx]}layers_{node_string}nodes_Adam_Dropout.txt'
        all_res_seed = []
        with open(fname,'r') as f:
            for line in f.readlines():
                all_res_seed.append(float(line))
            int_list.append(max(all_res_seed))
            # int_list.append(all_res_seed[-1])
    
    dropout_test_accs_layers_mean.append(np.mean(np.array(int_list)))
    dropout_test_accs_layers_std.append(np.std(np.array(int_list)))

index_max_dropout = dropout_test_accs_layers_mean.index(max(dropout_test_accs_layers_mean))
```

```python
perceptron_results_mean = []
perceptron_results_std = []

for seed in range(1,11):
    fname = f'Results/Classical_network_0.7trainsize_testaccs_{seed}seed_0layers_nodes_Adam.txt'

    int_list = []
    with open(fname,'r') as f:
        # int_list.append(float(f.readlines()[-1]))
        all_res_seed = []
        with open(fname,'r') as f:
            for line in f.readlines():
                all_res_seed.append(float(line))
            int_list.append(max(all_res_seed))
            # int_list.append(all_res_seed[-1])

perceptron_results_mean.append(np.mean(np.array(int_list)))
perceptron_results_std.append(np.std(np.array(int_list)))

# perceptron_results_mean = np.array([perceptron_results_mean[i] for i in range(10)])
# perceptron_results_std = np.array([perceptron_results_std[i] for i in range(10)])

print(f'Perceptron results: {perceptron_results_mean[0]} +/- {perceptron_results_std[0]}')
# print(f'Best config {node_list[index_max]} | result: {cnn_test_accs_layers_mean[index_max]} +/- {cnn_test_accs_layers_std[index_max]}')
print(f'Best dropout {node_list[index_max_dropout]} | result: {dropout_test_accs_layers_mean[index_max_dropout]} +/- {dropout_test_accs_layers_std[index_max_dropout]}')
```
