```python
from matplotlib import pyplot as plt
import numpy as np
```

### Amplitude encoding

```python
batch_size = 64
stoc_step = 10
layers = 10

amp_ce_test_accs = []
```

```python
for l in [8,9,10]:
    amp_res = []

    for seed in range(1,11):
        fname_ce = f'Results/qnn/ampl/Cross_Entropy/test_acc_{l}layers_batchsize{batch_size}_stocstep{stoc_step}_{seed}seed_CE.txt'

        with open(fname_ce, 'r') as f:
            ce_vals = [float(i) for i in f.readlines()]
            # amp_res.append(max(ce_vals))
            amp_res.append(ce_vals[-1])
        
    amp_ce_test_accs.append(amp_res)
```

### Angle encoding

```python
batch_size = 64
stoc_step = 10
layers = 10

ang_ce_test_accs = []
ang_cedr_test_accs = []
```

```python
for l in [8,9,10]:
    amp_res_ce = []
    amp_res_cedr = []

    for seed in range(1,11):
        fname_ce = f'Results/qnn/angle/Cross_Entropy/test_acc_{l}layers_batchsize{batch_size}_stocstep{stoc_step}_{seed}seed_CE.txt'
        fname_cedr = f'Results/qnn/angle/Cross_Entropy_DataReuploading/test_acc_{l}layers_batchsize{batch_size}_stocstep{stoc_step}_{seed}seed_CE.txt'

        with open(fname_ce, 'r') as f:
            ce_vals = [float(i) for i in f.readlines()]
            # amp_res_ce.append(max(ce_vals))
            amp_res_ce.append(ce_vals[-1])
        
        with open(fname_cedr, 'r') as f:
            cedr_vals = [float(i) for i in f.readlines()]
            # amp_res_cedr.append(max(cedr_vals))
            amp_res_cedr.append(cedr_vals[-1])
    
    ang_ce_test_accs.append(amp_res_ce)
    ang_cedr_test_accs.append(amp_res_cedr)
```

```python
colors = ['#aadce0',
 '#ef8a47',
 '#376795',
 '#f7aa58',
 '#72bcd5',
 '#ffe6b7',
 '#e76254',
 '#1e466e',
 '#ffd06f',
 '#528fad']
```

```python
plt.rcParams.update({
    "text.usetex": False,
    "font.family": "serif",
    "font.sans-serif": "Palatino",})

data = [amp_ce_test_accs[0], ang_ce_test_accs[0], ang_cedr_test_accs[0]]

fig = plt.figure(figsize=(1.5*3.5,1.5*2.75))
ax = fig.add_axes([0, 0, 1, 1])
ax.set_title('8 Layers', size = 24)
bp = ax.boxplot(data)
ax.set_xticklabels(['Amplitude\n Encoding', 'Angle\n Encoding', 'Angle+DR'], size = 20)
ax.set_ylabel('Test accuracy', size = 20)
ax.set_ylim((0.8,0.9))
ax.set_yticks([0.8, 0.82, 0.84, 0.86, 0.88, 0.9])
ax.tick_params(axis='both', which='major', labelsize=20)
ax.grid()
# show plot
plt.show()
fig.savefig('box_plot_8layer.pdf', bbox_inches = 'tight')
```
