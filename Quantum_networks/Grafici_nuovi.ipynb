```python
from matplotlib import pyplot as plt
import numpy as np
```

### Amplitude encoding

```python
batch_size = 64
stoc_step = 10
layers = 10

amp_ce_test_accs_mean = []
amp_ce_test_accs_std = []
```

```python
for l in range(1,layers+1):
    amp_res = []

    for seed in range(1,11):
        fname_ce = f'Results/qnn/ampl/Cross_Entropy/test_acc_{l}layers_batchsize{batch_size}_stocstep{stoc_step}_{seed}seed_CE.txt'

        with open(fname_ce, 'r') as f:
            ce_vals = [float(i) for i in f.readlines()]
            # amp_res.append(max(ce_vals))
            amp_res.append(ce_vals[-1])
        
    amp_ce_test_accs_mean.append(np.mean(amp_res))
    amp_ce_test_accs_std.append(1.96*np.std(amp_res))

# amp_ce_test_accs_mean, amp_ce_test_accs_std
```

### Angle encoding

```python
batch_size = 64
stoc_step = 10
layers = 10

ang_ce_test_accs_mean = []
ang_ce_test_accs_std = []

ang_cedr_test_accs_mean = []
ang_cedr_test_accs_std = []
```

```python
for l in range(1,layers+1):
    amp_res_ce = []
    amp_res_cedr = []

    for seed in range(1,11):
        fname_ce = f'Results/qnn/angle/Cross_Entropy/test_acc_{l}layers_batchsize{batch_size}_stocstep{stoc_step}_{seed}seed_CE.txt'
        fname_cedr = f'Results/qnn/angle/Cross_Entropy_DataReuploading/test_acc_{l}layers_batchsize{batch_size}_stocstep{stoc_step}_{seed}seed_CE.txt'

        with open(fname_ce, 'r') as f:
            ce_vals = [float(i) for i in f.readlines()]
            # amp_res_ce.append(max(ce_vals))
            amp_res_ce.append(ce_vals[-1])
        
        with open(fname_cedr, 'r') as f:
            cedr_vals = [float(i) for i in f.readlines()]
            # amp_res_cedr.append(max(cedr_vals))
            amp_res_cedr.append(cedr_vals[-1])
    
    ang_ce_test_accs_mean.append(np.mean(amp_res_ce))
    ang_ce_test_accs_std.append(1.96*np.std(amp_res_ce)/np.sqrt(10))

    ang_cedr_test_accs_mean.append(np.mean(amp_res_cedr))
    ang_cedr_test_accs_std.append(1.96*np.std(amp_res_cedr)/np.sqrt(10))

# ang_cedr_test_accs_mean , ang_cedr_test_accs_std
# max(ang_cedr_test_accs_mean)
```

### Angle encoding - Dropout

```python
batch_size = 64
stoc_step = 10
layers = 8

ang_ce_test_accs_dropout_mean = []
ang_ce_test_accs_dropout_std = []

ang_cedr_test_accs_dropout_mean = []
ang_cedr_test_accs_dropout_std = []
```

```python
for l in range(1,layers+1):
    amp_res_ce = []
    amp_res_cedr = []

    for seed in range(1,11):
        fname_ce = f'Results/dropout/angle/Cross_Entropy/test_acc_{l}layers_batchsize{batch_size}_stocstep{stoc_step}_{seed}seed_CE.txt'
        fname_cedr = f'Results/dropout/angle/Cross_Entropy_DataReuploading/test_acc_{l}layers_batchsize{batch_size}_stocstep{stoc_step}_{seed}seed_CE.txt'

        with open(fname_ce, 'r') as f:
            ce_vals = [float(i) for i in f.readlines()]
            # amp_res_ce.append(max(ce_vals))
            amp_res_ce.append(ce_vals[-1])
        
        with open(fname_cedr, 'r') as f:
            cedr_vals = [float(i) for i in f.readlines()]
            # amp_res_cedr.append(max(cedr_vals))
            amp_res_cedr.append(cedr_vals[-1])
    
    ang_ce_test_accs_dropout_mean.append(np.mean(amp_res_ce))
    ang_ce_test_accs_dropout_std.append(1.96*np.std(amp_res_ce)/np.sqrt(10))

    ang_cedr_test_accs_dropout_mean.append(np.mean(amp_res_cedr))
    ang_cedr_test_accs_dropout_std.append(1.96*np.std(amp_res_cedr)/np.sqrt(10))

# ang_cedr_test_accs_dropout_mean , ang_cedr_test_accs_dropout_std
# max(ang_cedr_test_accs_dropout_mean)
```
